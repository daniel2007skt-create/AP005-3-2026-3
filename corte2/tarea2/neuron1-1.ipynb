# Práctica 2: Neurona Artificial con Error Cuadrático (neuron1.1)

## 1. Introducción y Modelo Matemático
En esta práctica se conserva el modelo básico de neurona lineal con una entrada y un peso:
$$y = w \cdot x$$

La diferencia fundamental respecto a la práctica anterior radica en la función de costo. Se implementa el **error cuadrático** para penalizar de forma más efectiva las desviaciones grandes entre la predicción y el valor objetivo:
$$E = (y - t)^2$$

Mediante la regla de la cadena, se calcula la derivada parcial del error con respecto al peso, obteniendo la dirección necesaria para aplicar el algoritmo de **descenso por gradiente**:
$$\frac{\partial E}{\partial w} = 2(y - t)x$$

```python
import time

inputs = [1, 2, 3, 4]
targets = [2, 4, 6, 8]

w = 0.1
epochs = 10
learning_rate = 0.1
pause_time = 0.1

def predict(i):
    return w * i

for epoch in range(epochs):
    predictions = [predict(i) for i in inputs]
    errors = [(p - t)**2 for p, t in zip(predictions, targets)]
    cost = sum(errors) / len(targets)
    
    errors_d = [2 * (p - t) for p, t in zip(predictions, targets)]
    weight_d = [e * i for e, i in zip(errors_d, inputs)]
    
    gradient = sum(weight_d) / len(weight_d)
    w -= learning_rate * gradient
    time.sleep(pause_time)

test_inputs = [5, 6]
test_targets = [10, 12]
test_predictions = [predict(i) for i in test_inputs]

for i, t, p in zip(test_inputs, test_targets, test_predictions):
    print(f"Entrada: {i} | Target: {t} | Prediccion: {p:.4f}")
```

## 2. Interpretación de Resultados
- **Optimización por Gradiente:** Al actualizar el parámetro utilizando la regla $w = w - \eta \cdot \text{gradiente}$, el peso avanza de manera iterativa reduciendo el error cuadrático medio en cada época.
- **Validación del Modelo:** Tras 10 épocas de entrenamiento, las entradas de prueba ($5$ y $6$) arrojan predicciones extremadamente cercanas a sus objetivos esperados ($10$ y $12$), comprobando que el algoritmo de optimización funciona correctamente.
