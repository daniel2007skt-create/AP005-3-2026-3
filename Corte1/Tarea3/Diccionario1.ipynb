# define la lista con los nombres de los estudiantes

```python
names = ['Jenny', 'Alexus', 'Sam', 'Grace']
```

# define la lista con las alturas correspondientes

```python
heights = [61, 70, 67, 64]
```

# empareja las listas nombres y alturas en un objeto zip

```python
zipStudents = zip(names, heights)
```

# muestra la referencia del objeto zip creado

```python
print("zipStudents: ", zipStudents)
```

# genera el diccionario relacionando nombres y alturas por comprension

```python
students = {key:value for key, value in zip(names, heights)}
```

# imprime el diccionario de estudiantes

```python
print(students)
```

# define la lista de bebidas

```python
drinks = ["espresso", "chai", "decaf", "drip"]
```

# define la lista de miligramos de cafeina

```python
caffeine = [64, 40, 0, 120]
```

# empaqueta bebidas y cafeina en una variable zip

```python
zipped_drinks = zip(drinks, caffeine)
```

# imprime el objeto iterable de bebidas con cafeina

```python
print(zipped_drinks)
```

# transforma el iterable zip en un diccionario de bebidas

```python
drinks_to_caffeine = {key:value for key, value in zipped_drinks}
```

# imprime el diccionario de cafeina por bebida

```python
print(drinks_to_caffeine)
```

# define la lista con los titulos de canciones

```python
songs = ["Like a Rolling Stone", "Satisfaction", "Imagine", "What's Going On", "Respect", "Good Vibrations"]
```

# define la lista con los conteos de reproducciones

```python
playcounts = [78, 29, 44, 21, 89, 5]
```

# genera el diccionario relacionando canciones con reproducciones mediante comprension

```python
plays = {key:value for key, value in zip(songs, playcounts)}
```

# muestra en pantalla el diccionario recien creado

```python
print(plays)
```

# agrega el tema Purple Haze con una sola reproduccion

```python
plays.update({"Purple Haze": 1})
```

# modifica las reproducciones acumuladas para la cancion Respect

```python
plays.update({"Respect": 94})
```

# despliega el diccionario de reproducciones con sus actualizaciones

```python
print("After: ", plays)
```
